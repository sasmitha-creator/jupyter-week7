```python
print("Kurtosis:", value.kurt())
```

```python
Q1 = value.quantile(0.25)
Q3 = value.quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = value[(value < lower) | (value > upper)]
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower)
print("Upper Limit:", upper)
print("Number of Outliers:", len(outliers))
print("Outliers:")
print(outliers)
```

```python
cv = (value.std() / value.mean()) * 100

print("Coefficient of Variation:", cv)
```

```python
mean = value.mean()
median = value.median()
print("Mean:", mean)
print("Median:", median)
if mean > median:
    print("Distribution is positively skewed")
elif mean < median:
    print("Distribution is negatively skewed")
else:
    print("Distribution is approximately symmetrical")
```
